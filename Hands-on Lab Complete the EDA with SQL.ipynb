```python
import csv, sqlite3 
import pandas as pd
```

```python
con = sqlite3.connect("my_data1.db")
```

```python
%sql sqlite:///my_data1.db
```

```python
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/labs/module_2/data/Spacex.csv"
```

```python
df = pd.read_csv(url)
```

```python
df.head()
```

```python
df.to_sql("SPACEXTBL", con, if_exists='replace', index=False, method='multi')
```

```python
import prettytable 
prettytable.DEFAULT = 'DEFAULT'
```

```python
%sql SELECT DISTINCT Launch_Site FROM SPACEXTBL;
```

```python
%sql SELECT * FROM SPACEXTBL WHERE Launch_Site LIKE 'CCA%' LIMIT 5;
```

```python
%sql SELECT SUM(PAYLOAD_MASS__KG_) AS Total_Payload_Mass FROM SPACEXTBL WHERE Customer = 'NASA (CRS)';
```

```python
%sql SELECT AVG(PAYLOAD_MASS__KG_) AS Average_Payload_Mass FROM SPACEXTBL WHERE Booster_Version = 'F9 v1.1';
```

```python
%sql SELECT MIN(Date) AS First_Successful_Ground_Landing FROM SPACEXTBL WHERE Landing_Outcome = 'Success (ground pad)';
```

```python
%sql SELECT Booster_Version FROM SPACEXTBL WHERE Landing_Outcome = 'Success (drone ship)' AND PAYLOAD_MASS__KG_ > 4000 AND PAYLOAD_MASS__KG_ < 6000;
```

```python
%sql SELECT Mission_Outcome, Count(Mission_Outcome) AS Total_Outcomes FROM SPACEXTBL GROUP BY Mission_Outcome;
```

```python
%sql SELECT * FROM SPACEXTBL WHERE Launch_Site LIKE 'CCA%' LIMIT 5;
```

```python
%sql SELECT Booster_Version FROM SPACEXTBL WHERE PAYLOAD_MASS__KG_ = (SELECT MAX(PAYLOAD_MASS__KG_) FROM SPACEXTBL);
```

```python
%sql SELECT substr(Date, 6, 2) AS MONTH, Landing_Outcome, Booster_Version, Launch_Site FROM SPACEXTBL WHERE Landing_Outcome = 'Failure (drone ship)' AND substr(Date, 1, 4) = '2015';
```

```python
%sql SELECT Landing_Outcome, COUNT(Landing_Outcome) AS OUTCOME_COUNT FROM SPACEXTBL WHERE Date BETWEEN '2010-06-04' AND '2017-03-20' GROUP BY Landing_Outcome ORDER BY OUTCOME_COUNT DESC;
```
